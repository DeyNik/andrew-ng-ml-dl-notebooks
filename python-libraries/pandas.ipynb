## Dataframe
A 2d tabular data structure with labellednrows and columns.  
- Label based indexing
- fast vectorized operations 
- built on numpy

```python
data = {
    'Name':['Oscar Piastri','Arvid','Gorge Rusell'],
    'Age' :[27,18,28],
    'Country':['Italy','Britain','UK']
}

df=pd.DataFrame([[1,2,3],[3,4,5],[5,6,7]],columns=["A","B","C"])
print(df)

df=pd.DataFrame(data)
print(df)

df = pd.read_csv("customers-100.csv")
print(df.head())

#load from excel using pd.read_excel()
```

```python

print(df.describe(), end='\n\n')

print(df.nunique(), end='\n\n')

print(df['City'].unique()[:5], end='\n\n')

print(df.info(), end='\n\n')

print(df.shape, end='\n\n')

print(df.tail(3))
```

```python
#df.loc[rows, columns]
print(df.loc[0])

print(df.loc[[0,1,5]])

print(df.loc[8:10])

print(df.loc[5:6,["City","Country"]])
```
