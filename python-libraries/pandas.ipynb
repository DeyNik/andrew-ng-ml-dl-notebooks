## Add/Remove Columns 

```python
employee = pd.DataFrame([[1,'Fanny'],[2,'Luca'],[3,'Kimmy']], columns=['Id','Name'])

print(employee,end='\n\nAfter\n\n')

employee['Age']=[25,30,99]

print(employee,end='\n\nAfter\n\n')

employee.rename(columns={'Name':'First Name'}, inplace=True)

print(employee,end='\n\nAfter\n\n')

import numpy as np
employee['Salary']= np.where(employee['Age']>30, 200000, 400000)
employee['EmpId']=[12,67,23]
print(employee)

```

```python
employee.drop(columns=['EmpId'], inplace=True)
print(employee)
```

```python
employee['Revenue']= employee['Salary']*100
employee
```

```python
employee['PaymentType'] = employee['Salary'].apply(lambda x:'Cash' if x<300000 else ('Online' if x>300000 else ''))
employee
```

```python
def calculate_pf(row):
    if row['Salary']>300000:
        return row['Salary']*(30/100)
    elif row['Salary']<300000:
        return row['Salary']*(10/100)
    
employee['Pf']= employee.apply(calculate_pf, axis=1) #1 is rows

employee
```

## Merging and Concatenating Data

```python
Department = pd.DataFrame([[1,'Finance'],[2,'Tech'],[3,'HR']], columns=['EmpId', 'DName'])

#Joins!
emp_new=pd.merge(employee, Department, left_on='Id', right_on='EmpId', how='inner')
emp_new
```

```python
tech = emp_new[emp_new['DName']=='Tech'].copy()
finance = emp_new[emp_new['DName']=='Finance'].copy()

new_df = pd.concat([tech, finance])
new_df
```
