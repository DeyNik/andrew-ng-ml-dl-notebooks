# MatPlotLib 
For data visualization

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
x=np.array([2023, 2024, 2025, 2026])
y= np.array([15,25,30,20])
z= np.array([23,45,68,88])
w= np.array([34,56,78,44])
#plt.plot(x,y)

line_style= dict(marker=".", markersize=20, 
         markerfacecolor="violet",
         markeredgecolor="purple",
         linestyle="dashdot", linewidth=2)

plt.plot(x,y, color="lightgreen" ,**line_style, label='s1')

plt.plot(x,z, color="yellow", **line_style)

plt.plot(x,w,color="pink",**line_style, label='s3')


#add labels
plt.title("Class size", fontsize=20, family="Arial", fontweight="bold", color="orange" )

plt.xlabel("Year", fontsize=15, family="Arial", fontweight="bold", color="slateblue")
plt.ylabel("Students", fontsize=15, family="Arial", fontweight="bold",color="slateblue" )


#custmize ticks
plt.tick_params(axis="both", colors="slateblue")
plt.legend()

plt.figure(figsize=(5,3),dpi=300)

plt.show()

```

### Grid Lines

```python
x=[1,2,3,4,5]
y=[5,10,15,20,25]

plt.grid(axis="y", linewidth=2, color="lightblue", linestyle="dashed")
plt.plot(x,y)
plt.show()
```

### Bar Chart

```python
categories = np.array(["Grains","Fruit","Vegetables","Protien","Berry", "Sweets"])
values = np.array([4,3,2,5,3,1])

#plt.bar(categories , values, color="skyblue")
bars =plt.barh(categories , values, color="skyblue")
bars[0].set_hatch('+')
bars[3].set_hatch('/')

plt.title("Daily Consumption")
plt.xlabel("Food")
plt.ylabel("Quantity")

plt.show()
```

### Pie Chart

```python
categories = ['Freshman','Sophomores','Juniors','Seniors']
values= np.array([300,250,275,225])

colors=["violet","coral","lightblue","lightgreen"]
plt.pie(values, labels=categories, 
        autopct="%1.2f%%", colors=colors,
        explode=[0,0,0.1,0],
        shadow=True,
        startangle=90)


plt.title("College Demography")



plt.show()
```

### Scatter Plot
For correlations , like study hours vs test scores

```python
x1= np.array([0,1,2,3,4,2,3,7,8,1])
y1 = np.array([55,78,99,40,99,69,87,99,45, 78])
x2= np.array([7,7,2,5,6,6,9,7,8,1])
y2 = np.array([53,88,99,70,89,99,77,99,85, 78])

plt.scatter(x1,y1, color="lightblue", alpha=0.5, s=50, label="Class A")
plt.scatter(x2,y2, color="red", alpha=0.5, s=50, label="Class B")

plt.legend()
plt.xlabel("Hours Studied")
plt.ylabel("Grade")
plt.title("Test Scores")

plt.show()
```

### Histogram
Distribution of quantative data

```python
scores= np.random.normal(loc=80, scale=10, size=100)
scores =np.clip(scores,0,100)
plt.hist(scores, bins =20, color="pink",edgecolor="black")

plt.show()
```

### Subplot
the entire figure, and Ax is a single plot

```python
x=np.random.normal(loc=10,size=5)

figures, axes = plt.subplots(2,2)
axes[0,0].plot(x,x*2, color="blue")
axes[0,0].set_title("x*2")

axes[0,1].plot(x,x/2, color="red")
axes[0,1].set_title("x/2")

axes[1,0].plot(x,pow(x,4),color="green")
axes[1,0].set_title("pow(x,4)")

plt.tight_layout()
plt.show()
```
