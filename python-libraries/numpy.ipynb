# Numpy Library
- Multidimensional array library

### why over lists? 
- As **numpy is very fast**. It **uses fixed types**. If we store 5 as a number, numpy cast it into Int64 by default. You can customize this to Int16, Int 8.
- **List** is on the other hand, we **store a lot more information like** : Size, Reference Count, Object Type, Object value - which ultimately stores more than just Int64
- **Numpy iterates without type checking**, which in list by default isn't configured.
- **List is scattered around the memory**, by **numpy maintains contiguous allocation of memory**  This benefits SIMD vector processing that gets to perform computation one at a time , rather than one at a time, as everything is at one place. Also, cache is always complete and easier to utilize.

- **Difference** List just support insertion, deletion, appending , concatenation etc. Numpy supports a lot more. Example : multiply arrays made using numpy directly.

### Slicing
- [start:stop:step] (goes till step-1)
- for 2d aarrays : **[row selection , column selection]**
- for 2d arrays put
    - [1,2] for specific,
    - [:,:] for everything,
    - [0:2,:] a slice
    - [::-1,:] reverse

```python
# get a specific element [r,c]
print(b[1,-2], b[1,-1])
print(b[0,3])

#get rows and columns 
print(f"row 1 = {b[0,:]}")
print(f"column 2 = {b[:,1]}")
print(f"access 0.8,9 ={b[1:,1:3]}")
print(f"all values divisible by 3: {b[b%3==0]} ")
```

### The CURSE of Mutability - np.copy to the rescue

```python
a=np.array([1,2,3])

b=a
c= np.copy(a) #does deep copy

b[0] = 1000

print(a,b,c)
```

```python
c=a.view() #creates similar looking object but changes affect both
c[0]=2
print(c,a)
```

### Statistics

```python
stats = np.array([[1,2,3],[4,5,6]])

print("original array\n", stats, end='\n\n')
print("Min: ",np.min(stats))
print("Max in axis 1 (col): ", np.max(stats, axis=1))
print("Sum of all elements: ", np.sum(stats))
```

### Stacking Arrays

```python
v1= np.array([1,2,3,4])
v2 = np.array([4,5,6,7])

print("Vertical Stack:\n",np.vstack([v1,v2,v2,v1]))

print("\n\n Horizontal Stack: \n",np.hstack((v1,v2)))

print("\n\n concatenate arrays: \n", np.concatenate((v1,v2), axis=0))
```
