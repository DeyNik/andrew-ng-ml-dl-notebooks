# Model Prediction with Multiple Variables

The model's prediction with multiple variables is given by the linear model:

$$
f_{\mathbf{w},b}(\mathbf{x}) = w_0x_0 + w_1x_1 + \cdots + w_{n-1}x_{n-1} + b
$$

or, in vector notation:

$$
f_{\mathbf{w},b}(\mathbf{x}) = \mathbf{w} \cdot \mathbf{x} + b
$$

where $\cdot$ denotes the vector **dot product**.

To demonstrate the dot product, we will implement prediction using the scalar form followed by the vector form.

```python
def predict(x,w,b):
    p = np.dot(x,w)+b
    return p

#pre-defined values
b_init = 785.1811367994083
w_init = np.array([ 0.39133535, 18.75376741, -53.36032453, -26.42131618])
print(f"w_init shape: {w_init.shape}, b_init type: {type(b_init)}")

#get a row for training data
x_vec= X_train[0,:]
print(f"x_vec shape {x_vec.shape}, x_vec value: {x_vec}")

#make a prediction
f_wb = predict(x_vec, w_init, b_init)
print(f"f_wb shape {f_wb.shape}, prediction :{f_wb}")
```

# 4 Compute Cost With Multiple Variables

The cost function with multiple variables J(w, b) is:

J(w, b) = (1 / 2m) * Σ(i = 0 to m - 1) (f(w, b)(x(i)) - y(i))²

where

f(w, b)(x(i)) = w · x(i) + b

Unlike previous labs, w and x(i) are vectors rather than scalars, allowing the model to use multiple features.

```python
def compute_cost(X,y,w,b):
    m=X.shape[0]

    cost=0
    for i in X:
        f_wb= np.dot(X,w) + b
        cost= cost + (f_wb - y)**2
    
    j_wb = (cost)/(2*m)
    return j_wb

cost = compute_cost(X_train, y_train, w_init, b_init)
print(f'COst at optimal w : {cost}')
```
